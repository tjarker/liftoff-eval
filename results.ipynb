```python
import pandas as pd

import matplotlib.pyplot as plt

colors = [
    "#4C72B0",  # blue
    "#4C72B0",  # blue
    "#55A868",  # green
    "#4C72B0",  # blue
]

df = pd.read_csv("results.csv")
plt.figure(figsize=(7, 3))
bars = plt.bar(df["framework"], df["runtime"], color=colors)
plt.ylabel("Runtime (s)")
plt.xticks(rotation=45, ha="right")

# Write values above bars
for bar in bars:
  h = bar.get_height()
  plt.text(
    bar.get_x() + bar.get_width() / 2,
    h,
    f"{h:.2f}",
    ha="center",
    va="bottom",
    fontsize=9,
  )

plt.ylim(0, 400)
plt.tight_layout()
plt.savefig("runtimes.pdf")
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("results.csv")


plt.figure(figsize=(7, 3))
ax = plt.gca()

colors = [
    "#4C72B0",  # blue
    "#55A868",  # green
    "#C44E52",  # red
    "#8172B2",  # purple
]

bars = ax.bar(df.values, color=colors)

#ax.set_title("Simulation Time Breakdown 4x4 Constellation")
ax.set_ylabel("Time (s)")

ax.ticklabel_format(axis="y", style="plain")
ax.yaxis.get_offset_text().set_visible(False)

# set y axis to max 14.5
ax.set_ylim(0, 4.5)

ax.grid(axis="y", linestyle="--", linewidth=0.8, alpha=0.7)

for bar in bars:
    h = bar.get_height()
    ax.text(
        bar.get_x() + bar.get_width() / 2,
        h,
        f"{h:.2f}",
        ha="center",
        va="bottom",
        fontsize=10,
    )

plt.tight_layout()

#write to pdf
plt.savefig("performance_constellation.pdf", format="pdf", bbox_inches="tight")
```
